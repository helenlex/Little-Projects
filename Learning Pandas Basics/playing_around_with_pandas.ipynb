# Filtering rows

```python
# Breaking down the statement
# df[]
# [   (condition1) & (condition2)    ]
# each condition references the series as you would normally

file[(file["region"] == "NSW") & (file["quantity"] > 3)]
```

```python
file[(file["status"] == "Delivered") & (file["category"] == "Electronics")]
```

```python
# series.isin() function similar to in() function in SQL
# there are 2 round brackets as the function only accepts 1 argument

# breaking down the 4th condition
# ( condition 4 )
#  series.isin((string1, string2))

file[(file["status"] == "Delivered") & (file["category"] == "Electronics") 
     & (file["region"].isin(("VIC", "QLD")))]
```

```python
# filtering rows by row and column names

# breaking down the statement
# df.loc[     ]
# row_slice_start:row_slice_end, [columns]
file.loc[0:5, ["product", "status"]]
```

```python
# filtering by index

# breaking down the statement
# df.iloc[     ]
# row_slice_start:row_slice_end, [columns as index numbers]

file.iloc[0:5, [4, 9]]
```

# Creating columns

```python
# creating the column
file["total"] = file["quantity"] * file["unit_price"] * (1 - file["discount_pct"] / 100)

# checking the values
file["total"]
```

```python
# checking all columns and new column we created, to see if the calculation was accurate
# i manually calculated the first row in excel to see if it worked

file[["quantity", "unit_price", "discount_pct", "total"]]
```

```python
# creating a new column by extracting the "month" from another column

file["order_month"] = file["order_date"].dt.month

# checking both columns to see that the month was extracted correctly
file[["order_month", "order_date"]]
```

```python

# creating a new column called "Size", where depending on the "Total column", will show values of Large or Small
file["size"] = np.where(file["total"] > 1000, "Large", "Small")

#checking the column
file["size"]
```

# Addressing missing data

```python
# give me a sum of all records that have values of NA
file.isna().sum()
```

```python
# for any NA values in the column "discount_pct", replace with 0, and apply to the original dataframe
file.fillna({"discount_pct": 0}, inplace= True)
```

```python
# dropping rows that have na values in the "region" column

file.dropna(subset="region", inplace= True)

# checking how many rows where dropped
file
```

# Grouping and Aggregating

```python
# outputs a series
file.groupby("region")["total"].sum()
```

```python
# outputs a dataframe, pandas adds a new index
file.groupby("region")["total"].sum().reset_index()
```

```python
# grouping the category column based on the averages of the quantity column
file.groupby("category")["quantity"].mean()
```

```python
# two aggregations at once
file.groupby("product").agg(
    order_count = ("order_id", "count"),
    total_sales = ("total", "sum"),
)
```

```python
# grouping 2 columns together

file.groupby(["region", "status"])["total"].sum()
```

```python
# outputting a data frame
file.groupby(["region", "status"])["total"].sum().reset_index()
```

```python
# pivoting the dataframe
file.groupby(["region", "status"])["total"].sum().unstack()
```

# Sorting and ranking

```python
# for the column "total", sort by descending and give me the first 10 results
file["total"].sort_values(ascending= False).head(10)
```

```python
# for each group of "customer_id", calculate the total sales of each group.
# give me the top 5 spenders
file.groupby(["customer_id"])["total"].sum().sort_values(ascending= False).head(5).reset_index()
```

```python
# creating a pivot table
# total should be the values
# horizontal columns should be categories
# vertical columns should be region
# aggregation of totals is a sum
file.pivot_table(values= ["total"], columns= "category", index= ["region"], aggfunc= "sum")
```

```python
# creating a dict and creating a new data frame
lookup_dict = {"region" : ["NSW", "VIC", "QLD", "WA", "SA"],
             "state_name": ["New South Wales", "Victoria", "Queensland", "Western Australia", "South Australia"],
             "manager": ["Priya Nair", "Tom Henderson", "Mei Zhang", "Jack O'Connell", "Sofia Rossi" ]
             }

lookup_df = pd.DataFrame(lookup_dict)
```

```python
# left join
file.merge(right= lookup_df, how= "left", on= "region")
```
