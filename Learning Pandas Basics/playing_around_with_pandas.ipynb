# Selecting specific series

```python
# just like the previous example, double square brackets always return a dataframe
file[["order_id","product","quantity"]]
```

# Filtering rows

```python
# Breaking down the statement
# df[]
# [   (condition1) & (condition2)    ]
# each condition references the series as you would normally

file[(file["region"] == "NSW") & (file["quantity"] > 3)]
```

```python
file[(file["status"] == "Delivered") & (file["category"] == "Electronics")]
```

```python
# series.isin() function similar to in() function in SQL
# there are 2 round brackets as the function only accepts 1 argument

# breaking down the 4th condition
# ( condition 4 )
#  series.isin((string1, string2))

file[(file["status"] == "Delivered") & (file["category"] == "Electronics") 
     & (file["region"].isin(("VIC", "QLD")))]
```

```python
# filtering rows by row and column names

# breaking down the statement
# df.loc[     ]
# row_slice_start:row_slice_end, [columns]
file.loc[0:5, ["product", "status"]]
```

```python
# filtering by index

# breaking down the statement
# df.iloc[     ]
# row_slice_start:row_slice_end, [columns as index numbers]

file.iloc[0:5, [4, 9]]
```
